```python
import numpy as np
import matplotlib.pyplot as plt

# 1. PARÁMETROS DEL MODELO
mu = 8       # Promedio histórico de días lluviosos por mes
sigma = 2    # Desviación estándar
dias_mes = 30
n_iter = 10000  # Número de simulaciones Montecarlo

# 2. SIMULACIÓN MONTECARLO
# Generamos 10,000 valores aleatorios con distribución normal N(8, 2)
np.random.seed(42)  # Semilla para reproducibilidad
lluvias = np.random.normal(mu, sigma, n_iter)

# Aseguramos que los días lluviosos estén dentro del rango 0 a 30
lluvias = np.clip(lluvias, 0, dias_mes)

# 3. CÁLCULO DE LA PROBABILIDAD DE RETRASO
# Un retraso ocurre si X > 10
retrasos = np.sum(lluvias > 10)
prob_retraso = retrasos / n_iter
promedio_lluvias = np.mean(lluvias)

# 4. RESULTADOS
print("Simulación Montecarlo de retrasos por lluvias")
print(f"Iteraciones: {n_iter}")
print(f"Promedio de lluvias (días): {promedio_lluvias:.2f}")
print(f"Probabilidad de retraso (>10 días lluviosos): {prob_retraso:.3f} ({prob_retraso*100:.1f}%)")

# 5. GRÁFICOS
plt.figure(figsize=(12,5))

# Histograma de frecuencias
plt.subplot(1,2,1)
plt.hist(lluvias, bins=20, color='skyblue', edgecolor='black', density=True)
plt.axvline(10, color='red', linestyle='--', label='Límite de retraso (10 días)')
plt.title('Distribución simulada de días lluviosos')
plt.xlabel('Días lluviosos en el mes')
plt.ylabel('Frecuencia relativa')
plt.legend()

# Curva de probabilidad acumulada
plt.subplot(1,2,2)
lluvias_ordenadas = np.sort(lluvias)
prob_acumulada = np.arange(1, n_iter+1) / n_iter
plt.plot(lluvias_ordenadas, prob_acumulada, color='green')
plt.axvline(10, color='red', linestyle='--', label='Límite de retraso (10 días)')
plt.title('Probabilidad acumulada de días lluviosos')
plt.xlabel('Días lluviosos en el mes')
plt.ylabel('Probabilidad acumulada')
plt.legend()

plt.tight_layout()
plt.show()
```
