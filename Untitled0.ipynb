```python
import numpy as np
import pandas as pd
import yfinance as yf
```

```python
#Parámetros de usuario
TICKER = "AMZN"
YEARS = 10
NDAYS = 30              #días hábiles
NPATHS = 100_000        #número de escenarios
TARGET_UP = 1.08

#Descarga precios ajustados (Adj Close) últimos 10 años
ticker = yf.Ticker(TICKER)
end = pd.Timestamp.today().normalize()
start = end - pd.DateOffset(years=YEARS)
df = ticker.history(start=start.strftime("%Y-%m-%d"), end=end.strftime("%Y-%m-%d"), interval="1d")
if df.empty:
    raise RuntimeError("No se obtuvieron datos; verifica conexión o ticker.")
prices = df["Close"].dropna()   # usar 'Close' o 'Adj Close' según prefieras

#Precio actual (último cierre)
S0 = float(prices.iloc[-1])

#Calculo rendimientos logarítmicos diarios y parámetros
logret = np.log(prices / prices.shift(1)).dropna()
mu_daily = logret.mean()                # rendimiento medio diario
sigma_daily = logret.std(ddof=1)        # volatilidad diaria (sample std)

#Convertir a escala diaria (ya están diarios) - para GBM usamos mu y sigma diarios
# 3) Simular GBM: S_t = S0 * exp( (mu - 0.5*sigma^2) * dt + sigma * sqrt(dt) * Z )
dt = 1.0  # 1 día (usamos números de pasos = días)
steps = NDAYS

#Vectorizado para eficiencia
np.random.seed(42)  # reproducible; quita o cambia semilla si quieres
Z = np.random.normal(size=(NPATHS, steps))
drift = (mu_daily - 0.5 * sigma_daily**2) * dt
diffusion = sigma_daily * np.sqrt(dt)

#Generar log-returns para cada paso y camino
log_returns = drift + diffusion * Z
# sumar acumulado por fila y exponenciar:
log_sum = np.cumsum(log_returns, axis=1)
S_paths_T = S0 * np.exp(log_sum[:, -1])   # precio en T (día 30) por cada path

#Probabilidad de superar +8%
threshold = S0 * TARGET_UP
prob = np.mean(S_paths_T > threshold)

#Resultados resumen
result = {
    "ticker": TICKER,
    "S0": S0,
    "years_of_data": YEARS,
    "mu_daily": float(mu_daily),
    "sigma_daily": float(sigma_daily),
    "n_paths": NPATHS,
    "n_days": NDAYS,
    "threshold": threshold,
    "probability": float(prob)
}


import json
print(json.dumps(result, indent=2))

#percentiles y gráfico simple
import matplotlib.pyplot as plt
pctl = np.percentile(S_paths_T, [1,5,25,50,75,95,99])
print("Percentiles (1,5,25,50,75,95,99):", pctl)

plt.figure(figsize=(8,4))
plt.hist(S_paths_T, bins=200, density=True)
plt.axvline(S0, linestyle="--", label=f"S0={S0:.2f}")
plt.axvline(threshold, color="red", linestyle="-.", label=f"Threshold (+8%)={threshold:.2f}")
plt.title(f"Distribución de S_T en {NDAYS} días ({NPATHS:,} escenarios)")
plt.legend()
plt.tight_layout()
plt.show()
```
