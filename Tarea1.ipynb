El articulo que me gustaria compartir es sobre el megaproyecto de gas que amenaza directamente a las ballenas y la biodiversidad en las aguas mexicanas del Golfo de California.
Basicamente lo que esta pasando es que el proyecto Saguaro quiere exportar gas natural por el Golfo de California causando un impacto ambiental irremediable, este proyecto estaria desplazando y/o matando a las ballenas ya que este mar es parte de su ruta. Para recibir el gas y licuarlo, en Puerto Libertad se pretende construir una terminal de licuefacción. Esta megainfraestructura estaría generando ruido y quemando gas metano en mecheros permanentemente, alterando las condiciones actuales de la comunidad de pescadores. Finalmente, para llevar el gas licuado a Asia, enormes buques de 300 m de largo, llegarían al puerto para transportar el gas.
