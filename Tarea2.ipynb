```python
import sympy as sp

def primos(n):
    lista_primos = []
    for i in range(1, n):  # desde 1 hasta n-1
        if i == 1 or sp.isprime(i):  # considerar 1 y los primos
            lista_primos.append(i)
    return lista_primos
```

```python
print(primos(15))
```
